```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("..") / "data" / "indonesia"
df = pd.read_csv(DATA_DIR / "metadata.csv")

# v2 filename -> amendment type
AMENDMENT = {
    "DI_YOGYAKARTA_BANTUL_v2.txt": "Amendment",
    "DI_YOGYAKARTA_GUNUNGKIDUL_v2.txt": "Replacement",
    "JAWA_BARAT_BEKASI_KOTA_v2.txt": "Replacement",
    "JAWA_TENGAH_KARANGANYAR_v2.txt": "Replacement",
    "JAWA_TENGAH_SEMARANG_KOTA_v2.txt": "Replacement",
    "KEPULAUAN_BANGKA_BELITUNG_BANGKA_BARAT_v2.txt": "Additional",
    "KEPULAUAN_BANGKA_BELITUNG_BANGKA_SELATAN_v2.txt": "Additional",
    "NUSA_TENGGARA_BARAT_LOMBOK_TENGAH_v2.txt": "Replacement",
    "SUMATERA_SELATAN_OGAN_KOMERING_ULU_v2.txt": "Replacement",
}

# v1 files that have been replaced (no longer in force)
REPLACED = [
    "DI_YOGYAKARTA_GUNUNGKIDUL.txt",
    "JAWA_BARAT_BEKASI_KOTA.txt",
    "JAWA_TENGAH_KARANGANYAR.txt",
    "JAWA_TENGAH_SEMARANG_KOTA.txt",
    "NUSA_TENGGARA_BARAT_LOMBOK_TENGAH.txt",
    "SUMATERA_SELATAN_OGAN_KOMERING_ULU.txt",
]

df["amendment"] = df["filename"].map(AMENDMENT)   # blank if not listed
df["in_force"] = ~df["filename"].isin(REPLACED)

# Check that every hardcoded filename exists in the metadata
missing = (set(AMENDMENT) | set(REPLACED)) - set(df["filename"])
if missing:
    print("WARNING: not found in metadata:", sorted(missing))

df.to_csv(DATA_DIR / "metadata.csv", index=False)
```
