```python
# first just make sure the data is cleaner first
from pathlib import Path

base_dir = Path("../../../data/raw/localpolicies")

def is_number_only(line: str) -> bool:
    return line.strip().isdigit()

for file_path in base_dir.glob("*.txt"):
    original = file_path.read_text(encoding="utf-8").splitlines(keepends=True)

    filtered = [
        line for line in original
        if not is_number_only(line)
    ]

    # Only write if changed
    if filtered != original:
        file_path.write_text("".join(filtered), encoding="utf-8")
        print(f"Updated: {file_path.name}")
```
