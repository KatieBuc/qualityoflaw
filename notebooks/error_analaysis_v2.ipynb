```python
from pathlib import Path
import pandas as pd

def get_project_root():
    return Path().resolve().parent


df = pd.read_csv(get_project_root().parent / "HH+error_analysis_v2.csv", delimiter=";")
df = df[df['golden_label_update']=='x']

# df = df[df['golden_label_update']!='x'].dropna(subset='golden_label_update')
# with pd.option_context('display.max_rows', None):
#     display(df[['filename', 'indicator_id', 'golden_label_update']])
```

## AI Error analysis

```python
overall = (
    df.groupby("error_type")
      .size()
      .rename("count")
      .reset_index()
)

overall["percent"] = overall["count"] / len(df) * 100
overall
```

```python
summary = (
    df.groupby("dimension")
      .agg(
          n=("dimension","size"),
          FP=("error_type", lambda x: (x=="false_positive").sum()),
          FN=("error_type", lambda x: (x=="false_negative").sum()),
      )
)

summary
```

```python
indicator_summary = (
    df.groupby(["indicator_id","indicator_value"])
      .agg(
          n=("indicator_id","size"),
          FP=("error_type", lambda x: (x=="false_positive").sum()),
          FN=("error_type", lambda x: (x=="false_negative").sum())
      )
      .reset_index()
)

indicator_summary = indicator_summary.sort_values("n",ascending=False)
indicator_summary
```
