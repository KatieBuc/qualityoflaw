```python
import pandas as pd

# Columns that uniquely identify an example
merge_keys = [
    "fullname",
    "filename",
    "indicator_id",
]

# ------------------------------------------------------------------
# 1. Bring human review columns onto the new model output
# ------------------------------------------------------------------

human_cols = merge_keys + [
    "golden_label_update",
    "explanation",
]

df_v3 = df_v3.merge(
    df[human_cols],
    on=merge_keys,
    how="left",
)

# ------------------------------------------------------------------
# 2. Compare old vs new errors
# ------------------------------------------------------------------

old_errors = set(map(tuple, df[merge_keys].to_numpy()))
new_errors = set(map(tuple, df_v3[merge_keys].to_numpy()))

corrected = old_errors - new_errors      # error disappeared
new = new_errors - old_errors            # brand new error
persisting = old_errors & new_errors     # still an error

print(f"Old errors:        {len(old_errors)}")
print(f"New model errors:  {len(new_errors)}")
print(f"Corrected errors:  {len(corrected)}")
print(f"Persisting errors: {len(persisting)}")
print(f"New errors:        {len(new)}")

# ------------------------------------------------------------------
# 3. Create a summary dataframe
# ------------------------------------------------------------------

summary = pd.DataFrame({
    "Metric": [
        "Old errors",
        "New model errors",
        "Corrected by new model",
        "Persisting errors",
        "New errors introduced",
    ],
    "Count": [
        len(old_errors),
        len(new_errors),
        len(corrected),
        len(persisting),
        len(new),
    ]
})

print(summary)

# ------------------------------------------------------------------
# 4. (Optional) Label each row in df_v3
# ------------------------------------------------------------------

df_v3["_key"] = list(map(tuple, df_v3[merge_keys].to_numpy()))

df_v3["error_status"] = df_v3["_key"].map(
    lambda k: "Persisting error" if k in persisting else "New error"
)

df_v3.drop(columns="_key", inplace=True)

# ------------------------------------------------------------------
# 5. Save
# ------------------------------------------------------------------

df_v3.to_csv("df_v3_with_human_review.csv", index=False)
summary.to_csv("error_summary.csv", index=False)
```

```python
merge_keys = ["fullname", "filename", "indicator_id"]

old_errors = set(map(tuple, df[merge_keys].to_numpy()))
new_errors = set(map(tuple, df_v3[merge_keys].to_numpy()))

corrected = old_errors - new_errors
persisting = old_errors & new_errors
new = new_errors - old_errors

old_df = df.copy()
new_df = df_v3.copy()

old_df["_key"] = list(map(tuple, old_df[merge_keys].to_numpy()))
new_df["_key"] = list(map(tuple, new_df[merge_keys].to_numpy()))

corrected_df = old_df[old_df["_key"].isin(corrected)].copy()
persisting_df = new_df[new_df["_key"].isin(persisting)].copy()
new_df_only = new_df[new_df["_key"].isin(new)].copy()
```

```python
def summarise_by_indicator(df_):
    return (
        df_
        .groupby(["indicator_value", "indicator_id", "error_type"])
        .size()
        .unstack(fill_value=0)
        .assign(total=lambda x: x.sum(axis=1))
        .sort_values("total", ascending=False)
    )

corrected_summary = summarise_by_indicator(corrected_df)
persisting_summary = summarise_by_indicator(persisting_df)
new_summary = summarise_by_indicator(new_df_only)
```

```python
def investigate_indicator(indicator_id):
    cols = [
        "fullname",
        "filename",
        "indicator_id",
        "indicator_value",
        "error_type",
        "golden_label",
        "golden_label_update",
        "golden_evidence",
        "pred_label",
        "value",
        "pred_value",
        "explanation",
        "evidence",
        "rationale",
    ]

    persisting = (
        persisting_df[persisting_df["indicator_id"] == indicator_id]
        .copy()
        .assign(error_status="Persisting")
    )

    new = (
        new_df_only[new_df_only["indicator_id"] == indicator_id]
        .copy()
        .assign(error_status="New")
    )

    result = pd.concat([persisting, new], ignore_index=True)

    if result.empty:
        print(f"No persisting or new errors for indicator_id={indicator_id}")
        return result

    result["human_reviewed"] = result["golden_label_update"].notna()

    # Add blank column after golden_label
    result.insert(
        result.columns.get_loc("golden_label_update") + 1,
        "golden_evidence",
        ""
    )

    return result[
        ["error_status", "human_reviewed"] + cols
    ].sort_values(
        ["error_status", "error_type", "fullname", "filename"]
    )
```

```python
top20 = summarise_by_indicator(df_v3).head(20).reset_index()['indicator_id'].values
top20
```

```python
output_dir = get_project_root().parent

for ind_ in top20:
    review_df = investigate_indicator(ind_)
    review_df.to_csv(
        output_dir / f"error_analysis_{ind_}.csv",
        index=False
    )
```

```python
def diagnose_indicator(indicator_id):
    cols = [
        "fullname",
        "filename",
        "indicator_id",
        "indicator_value",
        "error_status",
        "error_type",
        "golden_label",
        "golden_label_update",
        "pred_label",
        "value",
        "pred_value",
        "explanation",
        "evidence",
        "rationale",
    ]

    result = (
        df_v3[df_v3["indicator_id"] == str(indicator_id)]
        .copy()
    )

    if result.empty:
        print(f"No current errors for indicator_id={indicator_id}")
        return result

    result["human_reviewed"] = result["golden_label_update"].notna()

    print(f"Indicator {indicator_id}")
    print(f"Current errors: {len(result)}")
    print()

    print("Status")
    print(result["error_status"].value_counts())
    print()

    print("Error type")
    print(result["error_type"].value_counts())
    print()

    print("Human reviewed")
    print(result["human_reviewed"].value_counts())
    print()

    return result[cols].sort_values(
        ["human_reviewed", "error_type", "error_status"],
        ascending=[False, True, True]
    )
```
